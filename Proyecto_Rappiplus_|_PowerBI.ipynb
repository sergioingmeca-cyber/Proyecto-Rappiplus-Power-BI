Para nuestro informe mediante Medidas DAX calculamos datos importantes para el informe como nuestro ticket promedio, costos totales, ventas totales y encontrar nuestro profit de la operacion de la compañia

En este notebook dejo las formulas de estas medidas:

* profit = SUM(orders_clean[monto_total]) - SUM(orders_clean[costo_total])
* costo_total = orders_clean[cantidad] * orders_clean[Costo Unitario]
* Ticket Promedio = SUM(orders_clean[monto_total]) / COUNT(orders_clean[id_pedido])  
* Costo Unitario = RELATED(catalog_clean[costo_unitario])
* Nombre Mes = FORMAT(orders_clean[fecha_hora_pedido], "MMMM")
